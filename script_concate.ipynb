```python
value_cols = [
    "description",
    "actual_area",
    "percent_of_base",
    "effective_area",
    "replacement_cost_new"
]
agg_map = {
    "description": "first",
    "actual_area": "sum",
    "percent_of_base": "sum",
    "effective_area": "sum",
    "replacement_cost_new": "sum"
}

grouped_df = (
    df3
    .groupby(["parcel_number", "page_save", "sub_area_type"], as_index=False)
    .agg(agg_map)
)
pivot_df = (
    grouped_df
    .set_index(["parcel_number", "page_save", "sub_area_type"])[value_cols]
    .unstack("sub_area_type")
)
sub_area_order = sorted(grouped_df["sub_area_type"].unique())

ordered_cols = []
for sub in sub_area_order:
    for col in value_cols:
        ordered_cols.append((col, sub))

pivot_df = pivot_df[ordered_cols]
pivot_df.columns = [
    f"{sub}_{col}"
    for col, sub in pivot_df.columns
]

pivot_df = pivot_df.reset_index()
merged_df = pd.merge(
    merged_df,
    pivot_df,
    on=["parcel_number", "page_save"],
    how="left"
)
cols = [c for c in merged_df.columns if c != "page_save"] + ["page_save"]
merged_df = merged_df[cols]
```

```python
df4 = pd.read_csv(r"D:\North Carolina\Cabarrus\Property\clean\us_nc_37025_data_prop_building-structural-elements_silver_20251127.csv")
df4
```

```python
# correct value columns
values_cols = [
    "detail_description",
    "point_value",
    "percentage"
]

# aggregate duplicates
agge_map = {
    "detail_description": "first",
    "point_value": "sum",
    "percentage": "sum"
}

grouped_df4 = (
    df4
    .groupby(["parcel_number", "page_save", "element"], as_index=False)
    .agg(agge_map)
)

# pivot to wide format
pivot_df4 = (
    grouped_df4
    .set_index(["parcel_number", "page_save", "element"])[values_cols]
    .unstack("element")
)

# enforce element-wise ordering
element_order = sorted(grouped_df4["element"].unique())

ordered_cols = []
for el in element_order:
    for col in values_cols:
        ordered_cols.append((col, el))

pivot_df4 = pivot_df4[ordered_cols]

# flatten column names
pivot_df4.columns = [
    f"{el}_{col}"
    for col, el in pivot_df4.columns
]

pivot_df4 = pivot_df4.reset_index()
```

```python
merged_df = pd.merge(
    merged_df,
    pivot_df4,
    on=["parcel_number", "page_save"],
    how="left"
)

# keep page_save last
cols = [c for c in merged_df.columns if c != "page_save"] + ["page_save"]
merged_df = merged_df[cols]
merged_df
```

```python
df5 = pd.read_csv(r"D:\North Carolina\Cabarrus\Property\clean\us_nc_37025_data_prop_building-bed-bath-half_silver_20251127.csv")
df5
```

```python
# correct value columns
valuess_cols = [
    "base_area",
    "upper_level",
    "lower_level"
]

# aggregate duplicates
aggee_map = {
    "base_area": "first",
    "upper_level": "sum",
    "lower_level": "sum"
}

grouped_df5 = (
    df5
    .groupby(["parcel_number", "page_save", "room_type"], as_index=False)
    .agg(aggee_map)
)

# pivot to wide format
pivot_df5 = (
    grouped_df5
    .set_index(["parcel_number", "page_save", "room_type"])[valuess_cols]
    .unstack("room_type")
)

# enforce element-wise ordering
element_order = sorted(grouped_df5["room_type"].unique())

ordered_cols = []
for el in element_order:
    for col in valuess_cols:
        ordered_cols.append((col, el))

pivot_df5 = pivot_df5[ordered_cols]

# flatten column names
pivot_df5.columns = [
    f"{el}_{col}"
    for col, el in pivot_df5.columns
]
pivot_df5 = pivot_df5.reset_index()
```

```python
merged_df = pd.merge(
    merged_df,
    pivot_df5,
    on=["parcel_number", "page_save"],
    how="left"
)

# keep page_save last
cols = [c for c in merged_df.columns if c != "page_save"] + ["page_save"]
merged_df = merged_df[cols]
merged_df
```
